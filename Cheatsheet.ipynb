### Button

```python
# PULL_UP Erklärung: bedeutet von 'Pin' über 'Schalter' zu 'GND'
    # Dabei ist der Pin auf HIGH, wenn der Schalter ungedrückt ist und 
    # auf LOW, wenn der Schalter gedrückt ist.

# PULL_DOWN Erklärung: bedeutet von 'Pin' über 'Schalter' zu '3V3'
    # Dabei ist der Pin auf LOW, wenn der Schalter ungedrückt ist und 
    # auf HIGH, wenn der Schalter gedrückt ist.


BUTTON_PIN = 2      # --> Anfangs Pins deffinieren.

button_pin = Pin(BUTTON_PIN, Pin.IN, Pin.PULL_UP)        # Dann muss der Pull-Up/Pull-Down bestimmt werden.
# button_pin = Pin(BUTTON_PIN, Pin.IN, Pin.PULL_DOWN)  



# Einfach: Für Led Pin, der über den Button Pin gesteuert wird. 
while True:
    led_pin.value(button_pin.value())       

# Komplex: Für Led Pin, wenn man es umdrehen will. Bei Button value = 0 --> Led value = 1 
while True:
    if button_pin.value() == 0:
        led_pin.on()
    else:
        led_pin.off()
```

### State_machine_1

```python
# Pins
PIN_BUTTON_OFF = 4
PIN_BUTTON_ON = 2
PIN_LED = 15

# Hardware
btn_off = Pin(PIN_BUTTON_OFF, Pin.IN, Pin.PULL_UP)
btn_on = Pin(PIN_BUTTON_ON, Pin.IN, Pin.PULL_UP)
led = Pin(PIN_LED, Pin.OUT)



# State Machine
STATE_OFF = 0
STATE_ON = 1
current_state = STATE_OFF

last_btn_off_state = btn_off.value()
last_btn_on_state = btn_on.value()

while True:
    btn_off_value = btn_off.value()
    btn_on_value = btn_on.value()
    
    # Transitions / Übergänge
    if current_state == STATE_OFF:
        if btn_on_value == 0 and btn_on_value != last_btn_on_state:
            current_state = STATE_ON
            print("off -> on")
            
    elif current_state == STATE_ON:
        if btn_off_value == 0 and btn_off_value != last_btn_off_state:
            current_state = STATE_OFF
            print("on -> off")

    # Aktionen
    if current_state == STATE_OFF:
        led.off()
    elif current_state == STATE_ON:
        led.on()
    
    last_btn_off_state = btn_off_value
    last_btn_on_state = btn_on_value


```

### State_machine_2   "Ampelmodel"

```python
# Pins
PIN_BUTTON = 4
PIN_LED_CAR_RED = 23
PIN_LED_CAR_GREEN = 21
PIN_LED_PED_RED = 21
PIN_LED_PED_GREEN = 19

# Hardware
btn1 = Pin(PIN_BUTTON, Pin.IN, Pin.PULL_UP)
led_car_red = Pin(PIN_LED_CAR_RED, Pin.OUT)
led_car_green = Pin(PIN_LED_CAR_GREEN, Pin.OUT)
led_ped_red = Pin(PIN_LED_PED_RED, Pin.OUT)
led_ped_green = Pin(PIN_LED_PED_GREEN, Pin.OUT)

# State Machine
STATE_CAR_GREEN = 0
STATE_ALL_RED_1 = 1
STATE_PED_GREEN = 2
STATE_ALL_RED_2 = 3
current_state = STATE_CAR_GREEN

last_button_state = btn1.value()
start_time = 0


# Pins definieren High oder Low
def all_leds_off():
    led_car_red.off()
    led_car_green.off()
    led_ped_red.off()
    led_ped_green.off()


while True:
    button1_value = btn1.value()
    
    # Transitions / Übergänge
    if current_state == STATE_CAR_GREEN:
        if button1_value == 0 and button1_value != last_button_state:
            current_state = STATE_ALL_RED_1
            print("car green -> all red 1")
            start_time = time.ticks_ms()                    # wenn eine zeitabhängiger Übergang ist

    elif current_state == STATE_ALL_RED_1:
        if time.ticks_diff(start_time, time.ticks_ms()) >= 1000:
            current_state = STATE_PED_GREEN
            print("all red 1 -> ped green")
            start_time = time.ticks_ms()

    elif current_state == STATE_PED_GREEN:
        if time.ticks_diff(start_time, time.ticks_ms()) >= 4000:
            current_state = STATE_ALL_RED_2
            print("ped green -> all red 2")
            start_time = time.ticks_ms()

    elif current_state == STATE_ALL_RED_2:
        if time.ticks_diff(start_time, time.ticks_ms()) >= 1000:
            current_state = STATE_CAR_GREEN
            print("all red 2 -> car green")

    # Aktionen
    if current_state == STATE_CAR_GREEN:
        all_leds_off()                              # Man muss immer alle bestimmen.
        led_car_green.on()
        led_ped_red.on()

    elif current_state == STATE_ALL_RED_1 or current_state == STATE_ALL_RED_2:
        all_leds_off()
        led_car_red.on()
        led_ped_red.on()

    elif current_state == STATE_PED_GREEN:
        all_leds_off()
        led_car_red.on()
        led_ped_green.on()
    
    last_button_state = button1_value
```
